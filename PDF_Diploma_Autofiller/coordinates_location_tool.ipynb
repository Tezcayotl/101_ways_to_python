````python
# Create an empty dummy PDF document
dummydoc = pymupdf.open()
dpno1 = dummydoc.new_page() # dpno1 states for dummydoc page number 1
# Save it to disk
dummydoc.save("dummy_document.pdf")
dummydoc.close()
````

Now A displayed "Lorem ipsum" legend will be printed at a position in the file
As a way to dinamically get the position, and other specs. 
ipywidgets **@interact** decorator will be added

````python

@interact(
    x_coord=(0,500),
    y_coord=(0,500),
)

def write_and_display(x_coord: int, y_coord: int):
    doc = pymupdf.open("dummy_document.pdf")
    page = doc[0]
    page.insert_text(
        pymupdf.Point(x_coord, y_coord),
        'Lorem Ipsum',
        fontsize=32,
        color=(0,0,0),
        fontname="helv",
    )
    doc.save("named_document.pdf")
    doc.close()
    IFrame("named_document.pdf", width=750, height=500)
````
